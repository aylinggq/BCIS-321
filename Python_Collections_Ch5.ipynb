```python
#Sara, a head TA is asked to write a program that can add and drop employees as the semester goes on. You have a dictionary to start made up of the current
#employees and their pay. You also have a list of previously terminated employees. Your task is to make sure that if you hire someone to add them to the
#dictionary. If someone is fired they need to be removed from the dictionary and added in the first position of the list, NOT the last.
#Ask your user if they want to add, remove, or view the current employees.
#Depending on what they say add, remove, or show the list and dictionary.
#If they are adding ask how much the new employee is paid and add that to the dictionary with their name.
#Here is your list and dictionary:
#SaraDictionary = {'Aubrey':15.25 , 'Sam':14.5, 'Lucia':10.75, 'Tyler':7.5}
#TerminatedEmployees = ['Benjamin', 'David']



SaraDictionary = {'Aubrey':15.25 , 'Sam':14.5, 'Lucia':10.75, 'Tyler':7.5}
TerminatedEmployees = ['Benjamin', 'David']

want = input("Do you want to add, remove, or view the current employees? ")

if want == "add":
  name = input("What is the name of the employee you want to add? ")

  if name in SaraDictionary:
    print(name, "is already in the dictionary")
  else:
    pay = float(input("What is the pay of the employee you want to add? "))
    SaraDictionary[name] = pay
    print(name, "has been added.")

elif want == "remove":
  name = input("What is the name of the employee you want to remove? ")

  if name in SaraDictionary:
    del SaraDictionary [name]
    TerminatedEmployees.insert(0, name)
    print(name, "has been removed and added to terminated list.")
  else:
    print(name, "is not in the dictionary.")

elif want == "view":
  print()
  print("Current Employees and Pay:")
  for employee, pay in SaraDictionary.items():
      print(employee, ":", pay)

  print()
  print("Terminated Employees:")
  print(TerminatedEmployees)
else:
    print("Invalid option. Please choose add, remove, or view.")
```
