```python
#You are a medical researcher running a study on a new drug. You are screening potential participants and you
#need to make sure you dont get any outliers. In particular, you want to flag anyone who is too tall/short or light/heavy.
#You need people who are in the "middle" so to speak to. Therefore, write a simple program to determine if a potential
#participant is eligible based on Weight, Height and Sex
#Men within 62 inches and 80 inches tall and between 150 and 280 lbs are needed for this study. Women between 59 and 71 inches
#and 100 and 180 lbs are needed for this study.
#Create flow logic to determine whether an inputted patient is within the thresholds. If they are an outlier, print a message
#indicating they are not eligible and tell them which thresholds they did not meet.
#In order to analyze data, we need to handle the many possible ways that these data could be entered. For example, sex could be
#entered into a textbox as "female", "f", "Female", 0 (female), 1 (male), "Male", "Man", "woman", or it could be mispelled.
#Use some combination of if/elif/else statements (including "or" and/or "and" criteria) and/or string formatting
#(e.g. converting case or examining only the first letter) to handle as many possibilities as you can think of to handle every type of input.
#For the height and weight, indicate that the user should enter numbers only. Using Python, check to see if a number is entered
#(i.e. not string/text data). If not, print a message stating that the user did not enter a number and ask them to start over.

my_gender=str(input("What is your gender?"))

if my_gender in("female"or"Woman"or"Female"or"woman"or"girl"or "girl"):
  gender= "female"
elif my_gender in ("male"or"Male"or"Man"or"man"or"boy"or"Boy"):
  gender="male"

my_height=input("What is your height in inches?")
try:
  height=float(my_height)
except:
  print("Invalid height.")

my_weight=input("What is your weight?")
try:
  weight=float(my_weight)
except:
  print("Invalid weight.")

outliers=[]

if gender=="male":
  if height<62 or height>80:
    outliers.append("height")
  if weight<150 or weight>280:
    outliers.append("weight")

elif gender=="female":
  if height<59 or height>71:
    outliers.append("height")
  if weight<100 or weight>180:
    outliers.append("weight")

if len(outliers)==0:
  print("Eligible for the study")
else:
  print("Not eligible in:", " and ".join(outliers))
```
