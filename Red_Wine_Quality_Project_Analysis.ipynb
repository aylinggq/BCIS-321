```python
import seaborn as sns
plt.subplot()
sns.countplot(x='quality', data=df, hue='quality', palette='dark:purple', legend=False)
plt.title('Distribution of Red Wine Quality')
plt.xlabel('Quality Score (3-8)')
plt.ylabel('Count')
# Interpretation: The dataset is unbalanced; most wines are rated 5 or 6, while excellent (8) or poor (3) wines are rare.
```

```python
plt.subplot()
sns.boxplot(x='quality', y='alcohol', hue='quality', data=df, palette='dark:purple')
plt.title('Alcohol Content by Wine Quality')
plt.xlabel('Quality Score')
plt.ylabel('Alcohol (%)')
# Interpretation: Higher quality wines (7 and 8) generally have a higher median alcohol content compared to lower quality wines.
```

```python
plt.subplot()
sns.scatterplot(x='volatile_acidity', y='citric_acid', hue='quality', data=df, palette='coolwarm', alpha=0.6)
plt.title('Volatile Acidity vs. Citric Acid')
plt.xlabel('Volatile Acidity (Acetic Acid)')
plt.ylabel('Citric Acid')
# Interpretation: There is a negative correlation between Volatile Acidity and Citric Acid, higher quality wines (darker blue) tend to cluster where volatile
# acidity is lower and citric acid is higher.
```

```python
plt.subplot()
correlation = df.corr()
sns.heatmap(correlation, annot=False, cmap='RdBu_r', center=0)
plt.title('Correlation Heatmap of Wine Features')
# Interpretation: This shows which variables move together. For instance,'fixed acidity' and 'pH' have a strong negative correlation (as expected),
# while 'alcohol' has the strongest positive correlation with 'quality'.
```
